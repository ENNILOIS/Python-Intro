## 11.1 Types of Errors

Three categories:
- **Syntax error** — This is when the code itself is Written Incorrectly, so Python cannot Understand It..

           SYNTAX = THE WAY THE CODE IS WRITTEN
- **Runtime error** — This is when code is Written Correctly, but something goes Wrong when Python tries to Run It..

            RUNRIME = PYTHON UNDERSTANDS THE CODE BUT CANNOT FIND / ACCESS WHAT U'RE ASKING FOR
- **Logic error** — This is when the code Runs Successfully, but the Answer or Output is not what you Intended..

            LOGIC = THE PROGRAM RUNS, BUT UR LOGIC / CALCULATION IS WRONG

## 11.3 else and finally Clauses

```python
# CLASS ACTIVITY 11.3a
# Add else and finally to this try/except block:
# - else: print a success message with the computed area
# - finally: print 'Calculation complete.'

def rectangle_area(length, width):
    try:
        if length <= 0 or width <= 0:
            raise ValueError('Dimensions must be positive.')
        area = length * width
    except ValueError as e:
        print(f'Error: {e}')
    # Add else and finally here:
    else:
        print(f"Success: Value of Length is multiplied by the Value of Width. \nArea = {area}")
    finally:
        print("CALCULATION COMPLETE.")

rectangle_area(5, 3)
print()
rectangle_area(-2, 4)
```

## 11.4 Raising Exceptions


```python
# CLASS ACTIVITY 11.4a
# Write a function validate_password(password) that raises:
# - ValueError if the password is shorter than 8 characters
# - ValueError if the password contains no digits
# Then call it and catch the errors.

def validate_password(password):
    pass  # your code here
    if len(password) < 8:
        raise ValueError ("Password must be at lease 8 characters.")
    if not any(char.isdigit() for char in password):
        raise ValueError ("Password must at least contain one digit. ")
    return password
try:
    validate_password('abc')
except ValueError as e:
    print(f'Error: {e}')

try:
    validate_password('abcdefgh')  # no digits
except ValueError as e:
    print(f'Error: {e}')

validate_password('secure123')   # should not raise
print('Password accepted.')
```

```python
# CLASS ACTIVITY 11.4b
# Write a function withdraw(balance, amount) that raises ValueError
# if amount <= 0 or if amount > balance.
# Test all three scenarios: valid withdrawal, negative amount, overdraft.

def withdraw(balance, amount):
    pass  # your code here
    if amount <= 0 or amount > balance:
        raise ValueError ("Amount must be greater than Zero and not more than the balance")
    return balance - amount

# Test 1: valid
print(withdraw(500, 100))   # 400

# Test 2: negative amount
try:
    withdraw(500, -50)
except ValueError as e:
    print(f'Error: {e}')

# Test 3: overdraft
try:
    withdraw(500, 600)
except ValueError as e:
    print(f'Error: {e}')
```

```python
# CLASS ACTIVITY 11.5b
# Add assert statements to validate the preconditions of this function:
# - items must be a non-empty list
# - discount_percent must be between 0 and 100

def calculate_discounted_total(items, discount_percent):
    # Add your asserts here:
    assert isinstance(items, list) and len(items) > 0, "Item must be a non-empty List."
    assert 0 <= discount_percent <= 100, "Discount_percent must be between 0 and 100"

    total = sum(items)
    return total * (1 - discount_percent / 100)

print(calculate_discounted_total([10, 20, 30], 10))   # 54.0
```
