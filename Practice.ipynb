```python
# CLASS ACTIVITY 10.3b
# Define a function is_even(number) that returns True if the
# number is even, False otherwise.
# Then use it in a loop to print whether each number 1-10 is even or odd.

# Your code here:
def is_even(numbers):
    if numbers % 2 == 0:
        return "True"
        
    return "False"

for i in range(1, 11):
    print(i, is_even(i))

print("-" * 25)

def even_or_odd(numbers):
    if numbers % 2 == 0:
        return "Even"
    
    return "Odd"

for n in range(1, 11):
    print(n, even_or_odd(n))
```

```python
# class Activity 10.4a
# Define a function send_message(recipient, message, urgent=False)
# If urgent is True, print: "[URGENT] To: reipient — message"
# Otherwise print: "To: recipient — message"
# Call it 3 times: once with urgent=True, twice without specifying urgent.

# Your code here:
# using Default Arguments 
def send_message(recipient, message, urgent=False):
    if urgent:
        print(f"[URGENT] To: {recipient} - {message}")
    else:
        print(f"To: {recipient} - {message}")
    print(f"Recipient Name: {recipient} / {message} / {urgent}")

send_message("Annie", "Life is Tiring, want to leave it", urgent=True)

print()

send_message("Bella", "I just wanted to know if i am wrong")

print()

send_message("Marbell", "Just Checking up on you")
```

```python
# CLASS ACTIVITY 10.4b
# Define a function calculate_discount(price, discount_percent=10)
# that returns the final price after the discount.
# Call it: once using the default 10%, once with 25%, once with 0%.

# Your code here:
def calculate_discount(price, discount_percent=10):
    return price - (price * discount_percent / 100)
    
    
print("Total Price =", calculate_discount(200000))
print("Total Price =", calculate_discount(400000, 25))
print("Total Price =", calculate_discount(600000, 0))
```
