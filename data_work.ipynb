```python
# Using 1D Array - A single sequence of values (Identified by Number of square Bracket [] used.)
daily_revenue = np.array([45000, 62000, 38000, 71000, 55000, 49000, 83000, 14000])

print(daily_revenue)
print('Type:', type(daily_revenue))  # Used to tell us what kind of Python Object something is.
print("Shape:", daily_revenue.shape) # Used to check the number of columns & rows in a data.
print('Dtype:', daily_revenue.dtype) # Used to tell us the data type of the elements inside a NumPy array
print("Ndim:", daily_revenue.ndim)   # Used to check Number Dimensions of the data.

print()

#  Using 2D Array - Rows and Colunms values (Identified by Number of square Bracket [[]] used.)
# Weekly revenue by product category (Electronics, Fashion, Food, Drinks)
weekly_revenue = np.array([
    [850.45, 120.01, 420.09, 560.11],  # Week 1
    [150.81, 910.06, 380.03, 200.07],  # Week 2
    [780.78, 550.55, 90.56, 100.21],   # Week 3
    [180.09, 100.54, 610.07, 131.59],  # Week 4
    [200.45, 155.55, 785.25, 171.51],  # Week 5
])

print(weekly_revenue)
print('Shape:', weekly_revenue.shape)
# print("Type:", type(weekly_revenue))
print('Ndim:', weekly_revenue.ndim)
print("Dtype:", weekly_revenue.dtype)
```

```python
# To Load Dataset in NumPy
import csv

# Load our orders dataset manually (we'll use pd.read_csv())
orders = []
with open("orders.csv", encoding='utf-8-sig') as f:
    reader = csv.DictReader(f)
    for row in reader:
        orders.append(row)

print(f"Loaded {len(orders)} orders")
print("Columns:", list(orders[0].keys()))
```

```python
# Extract total_amount as a NumPy array
# Some Rows have missing total_amount - skip those for now
import numpy as np

total_amounts = np.array([
    float(row["total_amount"])
    for row in orders
    if row["total_amount"] != ""
])

print("Array Shape:", total_amounts.shape)
print("First 11 values:", total_amounts[:11])
```

```python
# Apply 7.5% VAT to every order — no loop needed
vat = total_amounts * 0.075
total_with_vat = total_amounts + vat

print("First 6 Original Amounts: ", total_amounts[:6])
print("First 6 VAT Amounts:      ", vat[:6].round(2))
print("First 6 Total with VAT:   ", total_with_vat[:6].round(2))

print()
print()

# Convert Naira to USD (1 USD = 1,600 NGN)
usd_amounts = total_amounts / 1600

print("First 6 Amounts in USD:", usd_amounts[:6].round(2))
```

```python
print("Total Revenue:     NGN", f"{np.sum(total_amounts):,.0f}")
print("Average Order:     NGN", F"{np.mean(total_amounts):,.0f}")
print("Median Order:      NGN", f"{np.median(total_amounts):,.0f}")
print("Smallest Order:    NGN", f"{np.min(total_amounts):,.0f}")  # min = Minimum
print("Largest Order:     NGN", f"{np.max(total_amounts):,.0f}")  # max = Maximum
print("5th Deviation:     NGN", f"{np.std(total_amounts):,.0f}")
print("75th Percentile:   NGN", f"{np.percentile(total_amounts, 75):,.0f}")

print()
print()

# Why is the mean much higher than the median?
# If MEAN is Higher than Median - It is Called Outliers
# Let's find those extreme orders
mean_val = np.mean(total_amounts)
median_val = np.median(total_amounts)

print(f"Mean:         NGN {mean_val:,.0f}")
print(f"Median:       NGN {median_val:,.0f}")
print(f"Differance:   NGN {mean_val - median_val:,.0f}")

# Orders Above NGN 200,000 - The Outliers Pulling The Mean up
outliers = total_amounts[total_amounts > 200000]
print("Extreme Orders Above NGN 200,000:", outliers)
```

##### WHAT DOES .copy() DO IN PANDAS
**.copy** is a Pandas DataFrame/Series Method.

**.copy** is Used to make a Seperate Copy of a DataFrame or Series.

The Main reason you use it is to make Changes to the New DataFrame without accidentally changing the Original One.

  For Example:

     high_value_orders = df[df["total_amount] > 250000].copy()

  Here:
  
      df ⇀ your original DataFrame
      df[df["total_amount] > 25000] ⇀ selects only orders above 25,000
      .copy() ⇀ creates an independent copy of those selected rows

KEY: **.copy** is like Photocopying a Document.


```python
# Step 1: Filter to Electronics only
electronics = df[df["product_category"] == "Electronics"].copy()
print(f"Electronics Orders: {len(electronics)}")

# Step 2: Drop Rows with missing total_amount
electonics = electronics.dropna(subset=["total_amount"])
print(f"After Dropping Missing Amounts: {len(electronics)}")

print()

# Step 3: Summary By Payment Method
payment_summary = electronics.groupby("payment_method")["total_amount"].agg(
    order_count="count",
    total_revenue="sum",
    avg_order="mean"
).round(0)

payment_summary
```

##### UNDERSTANDING THE CODE BELOW:
**FROM CODE -- for method, row in payment_summary.itterrows():**

**What does .itterrows() mean?**

**.iterrows() Means:**

       Go through the DataFrame one Row at a Time. 
       .iterrows() takes them one row at a time.

**What is method?**

**method** 

       Gives the Row's Index (Your Payment Method)
**row**

      Gives all the Information in that Row.

      **KEY: **
      So for Bank Transfer:
          method is: Bank Transfer
      And:
          row contains:
          order_count    22
          total_revenue  1388965
          avg_order      63135
      That's why youcan do:
           row["total_revenue"] to get revenue.

**For the < & > Signs**

         **From Code -- {method:<20}** The < here does Not mean "Less than"
         It's part of Python's Formatting Syntax.

**<20 Means:**
          Put the Text on the Left and Give it a Space of 20 Characters.

      **FOR EXAMPLE:**
             f"{'card':<20}" -- Would make space for 20 Characters and put Card on the Left.
    
**>3 Means:** -- {int(row['order_count']):>3}

         Give me a Space of 3 Characters and put the Number on the right.
         This Helps the Numbers Line Up.

**>10,.0f** -- {row['total_revenue']:>10,.0f}

         There are actually three Formatting Instructions Here:
            - >10
                Right-Align the Number in a Space of 10 Characters. 
            - ,
                Add Commas to Large Numbers.
            - .0f
                Shows Zero Decimal PLaces.

**KEY:**

     Any where you see < any number & >any number Inside an f-String:
          - < any number Means
                  LEFT-ALIGN, THE NUMBER OF SPACES SPECIFIED.
          - > any number Means
                  RIGHT-ALIGN, THE NUMBER OF SPACES SPECIFIED.
        

```python
# Step 4: Format & Present
print("====== ELECTRONICS ORDERS BY PAYMENT METHOD ======")
for method, row in payment_summary.iterrows():
    print(f"  {method:<20} | Orders: {int(row['order_count']):>3} |"
          f"Revenue: NGN {row['total_revenue']:>10,.0f} | "
          f"Avg: NGN {row['avg_order']:>8,.0f}")
```

```python
# LOAD AND INSPECT DATA
import pandas as pd
import numpy as np

# LOAD RAW DATA
df = pd.read_csv("orders.csv", encoding='utf-8-sig')

print(f"Raw Dataset: {df.shape[0]} rows x {df.shape[1]} columns")
df.head(6)
```

```python
# TO MAKE A COPY OF THE DATA
# Always work on a copy - Never modify the original DataFrame
df_clean = df.copy()

print("Working Copy Created:", df_clean.shape)
print("Original Preserved:  ", df.shape)
```

```python
# TO FIX DATA TYPES

# unit_pric: strip the ₦ symbol, convert to float
print(f"Before - dtype:", df_clean["unit_price"].dtype)
print(f"Before - sample:", df_clean["unit_price"].head().tolist())

# Step 1: Remove the ₦ Symbol
df_clean["unit_price"] = df_clean["unit_price"].str.replace("₦", "", regex=False)

# Step 2: Convert to Numeric (errors='coerce' turns any remaining bad values to NaN)
df_clean["unit_price"] = pd.to_numeric(df_clean["unit_price"], errors='coerce')

print()
print("After - dtype:", df_clean["unit_price"].dtype)
print("After - sample:", df_clean["unit_price"].head().tolist())
print("Any New Missing Values?", df_clean["unit_price"].isnull().sum())
```

```python
# TO FIX MISSING VALUES

# Phone: Can't Recover a Missing Phone Number
# Fill it with a Placeholder - Keep the Row, Flags the Gap
df_clean["phone"] = df_clean["phone"].fillna("Not Provided")

print("Missing Phone Values After Fill:", df_clean["phone"].isnull().sum())
print("Sample Pnone Values:")
print(df_clean["phone"].value_counts().head())

print()
print()

# total_amount: this is our key revenue metric
# We cannot analyse revenue with missing amounts - drop those rows or calculate total_amount again

# before_shape = df_clean.shape
# df_clean = df_clean.dropna(subset=["total_amount"])
# after_shape = df_clean.shape

# print(f"Before Dropna: {before_shape[0]} rows")
# print(f"After Dropna:  {after_shape[0]} rows")
# print(f"Rows Removed:  {before_shape[0] - after_shape[0]}")
# print(f"Missing Total_Amount: {df_clean['total_amount'].isnull().sum()}")

# TO CALCULATE TOTAL_AMOUNT MISSING VALUES INSTEAD OF DROPPING
df_clean["total_amount"] = df_clean["total_amount"].fillna(
    df_clean["quantity"] * df_clean["unit_price"]
)

# THIS IS USED INCASE IF ANY COLUMN IN QUANTITY & UNIT_PRICE IS MISSING. IT SHOULD DROP THE TOTAL AMOUNT MISSING
df_clean = df_clean.dropna(subset=["total_amount"])

df_clean.head(20)
```

```python
print(df_clean["total_amount"].isnull().sum())
```

```python
# TO REMOVE DUPLICATES

before_shape = df_clean.shape
df_clean = df_clean.drop_duplicates()
after_shape = df_clean.shape

print(f"Before Drop_Duplicates: {before_shape[0]} rows")
print(f"After_Drop_Duplicates:  {after_shape[0]}")
print(f"Duplicated Removed:     {before_shape[0] - after_shape[0]}")

print()
print()

# Confirm Zero Duplicates Remaining
print(f"Remaining Duplicates: {df_clean.duplicated().sum()}")

# Also Check for Duplicate order_ids (same ID, slightly different rows)
print(f"Duplicate order_ids: ", df_clean.duplicated(subset=["order_id"]).sum())
```

```python
# TO FIX DATA TYPES

# TO  CONVERT DATE STRING TO DATETIME
# order_date: convert strings to Datetime
print("Before - dtype:", df_clean["order_date"].dtype)
print("Before - sample:", df_clean["order_date"].head(4).tolist())

df_clean["order_date"] = pd.to_datetime(df_clean["order_date"])

print()
print("After - dtype:", df_clean["order_date"].dtype)
print("After - sample:", df_clean["order_date"].head(4).tolist())

print()
print()

# TO CREATE NEW COLUMN DATA FROM ORDER_DATE
# Now we can Extract Data Components - Unlocked By the DataTime Conversion
df_clean["order_month"] = df_clean["order_date"].dt.month
df_clean["order_year"]  = df_clean["order_date"].dt.year

print("Orders Per Month:")
print(df_clean["order_month"].value_counts().sort_index())

print()

print("Orders Per Year:")
print(df_clean["order_year"].value_counts().sort_index())
print()       
print(df_clean["order_year"].unique())             # HOW TO USE *UNIQUE METHOD*
```

```python
# TO CLEAN STRING COLUMNS

# State: Standardise To Title Case
print("Before - Unique State Values:")
print(sorted(df_clean["state"].unique()))
print(f"Total Unique Value: {df_clean['state'].nunique()}")  # nunique - Gives the number of unique values.

print()
print()

# To Remove the inconsistency in name in the strings name column
df_clean["state"] = df_clean["state"].str.strip().str.title()

print("After - Unique State Values:")
print(sorted(df_clean["state"].unique()))
print(f"Total Unique Value: {df_clean['state'].nunique()}")
```
